```python
red = pd.read_csv(
    "dataset/wine+quality/winequality-red.csv",
    sep=";"
)

white = pd.read_csv(
    "dataset/wine+quality/winequality-white.csv",
    sep=";"
)

print("Both datasets loaded successfully!")
```

```python
print("Red Wine Dataset Shape:", red.shape)
print("White Wine Dataset Shape:", white.shape)

print("\nTotal Red Wine Records:", len(red))
print("Total White Wine Records:", len(white))
```

```python
plt.figure(figsize=(10,5))

sns.countplot(data=red, x="quality", color="red")

plt.title("Red Wine Quality Distribution")
plt.xlabel("Quality")
plt.ylabel("Count")

plt.show()
```

```python
plt.figure(figsize=(10,5))

sns.countplot(data=white, x="quality", color="skyblue")

plt.title("White Wine Quality Distribution")
plt.xlabel("Quality")
plt.ylabel("Count")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(red["alcohol"], bins=20, kde=True)

plt.title("Alcohol Distribution in Red Wine")
plt.xlabel("Alcohol Content")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(white["alcohol"], bins=20, kde=True)

plt.title("Alcohol Distribution in White Wine")
plt.xlabel("Alcohol Content")
plt.ylabel("Frequency")

plt.show()
```

```python
red["wine_type"] = "Red"
white["wine_type"] = "White"

wine = pd.concat([red, white], ignore_index=True)

print("Combined Dataset Shape:", wine.shape)

display(wine.head())
```

```python
print(
    wine.groupby("wine_type")["alcohol"].mean()
)
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(
    data=wine,
    x="wine_type",
    y="alcohol"
)

plt.title("Alcohol Content: Red vs White Wine")
plt.xlabel("Wine Type")
plt.ylabel("Alcohol Content")

plt.show()
```

```python
print(
    wine.groupby("wine_type")["quality"].mean()
)
```

```python
print("Average Alcohol Content:")
print(wine.groupby("wine_type")["alcohol"].mean())

print("\nAverage Wine Quality:")
print(wine.groupby("wine_type")["quality"].mean())

print("\nMedian Alcohol Content:")
print(wine.groupby("wine_type")["alcohol"].median())
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(data=wine, x="quality", y="alcohol", hue="wine_type")

plt.title("Alcohol Content vs Wine Quality")
plt.xlabel("Wine Quality")
plt.ylabel("Alcohol Content")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(data=wine, x="quality", y="volatile acidity", hue="wine_type")

plt.title("Volatile Acidity vs Wine Quality")
plt.xlabel("Wine Quality")
plt.ylabel("Volatile Acidity")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(data=wine, x="quality", y="sulphates", hue="wine_type")

plt.title("Sulphates vs Wine Quality")
plt.xlabel("Wine Quality")
plt.ylabel("Sulphates")

plt.show()
```
